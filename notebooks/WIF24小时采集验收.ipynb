# WIF24小时采集验收

本Notebook读取本地已有的历史验收摘要，不记录服务器路径。当前采集文件验收入口为tools/data/verify_self_collected.py。

摘要中的passed只表示当次限定检查通过，不替代当前数据连续性、逐笔完整性及回测认证。


```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display
root=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'data/server_acceptance').is_dir())
r=json.loads((root/'data/server_acceptance/2026-10-02/wif-acceptance.json').read_text())
print('通过：',r['passed'],'；错误数：',r['error_count'])
display(pd.DataFrame([{'频道':k,'消息数':v} for k,v in r['counts'].items()]))
display(pd.DataFrame(r['streams']).T)
print('深度回放：',r['depth_replay'])
print('质量事件：',r['quality_events'])
assert r['passed'] and r['error_count']==0
```
